**a. Describe the data. Number of rows and columns, what does one row represent.**

```python
# pip install geopandas requests pandas lxml
```

```python
import requests
import xml.etree.ElementTree as ET

WFS_URL = "https://gsavalik.envir.ee/geoserver/metsaregister/wfs"
LAYER = "metsaregister:eraldis"

params = {
    "service": "WFS",
    "version": "2.0.0",
    "request": "GetFeature",
    "typeNames": LAYER,
    "resultType": "hits"
}

response = requests.get(WFS_URL, params=params)
response.raise_for_status()

root = ET.fromstring(response.content)

rows = root.attrib.get("numberMatched")

params = {
    "service": "WFS",
    "version": "2.0.0",
    "request": "DescribeFeatureType",
    "typeNames": LAYER
}

response = requests.get(WFS_URL, params=params)
response.raise_for_status()

root = ET.fromstring(response.content)

columns = []

for element in root.iter():
    if element.tag.endswith("element"):
        name = element.attrib.get("name")
        if name:
            columns.append(name)

print("FOREST REGISTRY – METSAREGISTRI ERALDISED")
print("------------------------------------------")
print("Number of rows:", rows)
print("Number of columns:", len(columns))
print("\nColumn names:")

for col in columns:
    print(col)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from io import StringIO
import requests

WFS_URL = "https://gsavalik.envir.ee/geoserver/metsaregister/wfs"

params = {
    "service": "WFS",
    "version": "2.0.0",
    "request": "GetFeature",
    "typeNames": "metsaregister:eraldis",
    "propertyName": "peapuuliik_kood,pindala",
    "outputFormat": "csv",
}

response = requests.get(WFS_URL, params=params, timeout=120)
response.raise_for_status()

df = pd.read_csv(StringIO(response.text))

display(df.head())
print("Retrieved records:", len(df))
```

```python
df["pindala"] = pd.to_numeric(df["pindala"], errors="coerce")
df = df.dropna(subset=["peapuuliik_kood", "pindala"])
df = df[df["pindala"] > 0]

# Species distribution
plt.figure(figsize=(10, 5))
df["peapuuliik_kood"].value_counts().plot(kind="bar")
plt.title("Dominant Tree Species Distribution")
plt.xlabel("Species code")
plt.ylabel("Number of stands")
plt.tight_layout()
plt.show()

# Stand area distribution
plt.figure(figsize=(10, 5))
plt.hist(df["pindala"], bins=40, edgecolor="black")
plt.title("Forest Stand Area Distribution")
plt.xlabel("Area (ha)")
plt.ylabel("Number of stands")
plt.show()

# Area by species
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x="peapuuliik_kood", y="pindala")
plt.yscale("log")
plt.title("Forest Stand Area by Dominant Tree Species")
plt.xlabel("Species code")
plt.ylabel("Area (ha, log scale)")
plt.tight_layout()
plt.show()
```

**d. Check for data quality**

Missing values, duplicates, impossible or inconsistent values. Also watch out for outliers and anomalies that may require further investigation. Explain what you found and what you are going to do with the information.


```python
def count_features(cql_filter=None):
    """Count matching features without retrieving their attributes or geometry."""

    params = {
        "service": "WFS",
        "version": "2.0.0",
        "request": "GetFeature",
        "typeNames": LAYER,
        "resultType": "hits"
    }

    if cql_filter:
        params["CQL_FILTER"] = cql_filter

    response = requests.get(WFS_URL, params=params, timeout=120)
    response.raise_for_status()

    root = ET.fromstring(response.content)

    # WFS 2.0 normally returns numberMatched
    count = root.attrib.get("numberMatched")

    if count is None:
        count = root.attrib.get("numberOfFeatures")

    return int(count) if count not in (None, "unknown") else None

checks = {
    "Total forest stands": None,

    "Missing dominant species code":
        "peapuuliik_kood IS NULL",

    "Empty dominant species code":
        "peapuuliik_kood = ''",

    "Missing stand area":
        "pindala IS NULL",

    "Zero or negative stand area":
        "pindala <= 0",

    "Stand area below 0.01 ha":
        "pindala < 0.01",

    "Stand area above 100 ha":
        "pindala > 100"
}

results = []

for description, filter_expression in checks.items():
    try:
        number = count_features(filter_expression)
        results.append({
            "Quality check": description,
            "Number of stands": number
        })
    except Exception as e:
        results.append({
            "Quality check": description,
            "Number of stands": str(e)
        })

quality_report = pd.DataFrame(results)
display(quality_report)
```

**Answer:** Overall, the dataset appears to have good data completeness for the two examined variables. However, 4 stands have invalid or unusually small areas, while another 4 exceed 100 hectares and should be investigated as potential outliers.

This test should be run again on our specified test areas. Stands with missing species codes will be excluded from the classification reference data. The  stands with invalid areas will also be excluded. Large stands will not automatically be removed, as they may represent genuine forest compartments. Their values should be checked before deciding whether to retain them.

## Sentinel 2 data

Here we look at 1 example image of sentinel. This might not be the actual image we use in our project, or we might include more than one image.

**a. Describe the data. Number of rows and columns, what does one row represent?** <br>
The Sentinel 2 data consists of different satellite images. Each image is of a specific geographic area. In each image data is in the form of a raster, where each pixel represents small geographical region. The size of the pixel (how much land it covers in real life) is determined by the spatial resolution of the image. Each raster pixel contains the measured reflected spectral valuefor different spectral wavelenghts. Sentinel 2 collects multispectral data, which means that the electromagnetic spectrum has been divided into different bands. There are 13 bands in total. Each band is measured at different spatial resolutions:<br>
Bands and their resolutions:
* 60m pixel size: band1, band9,band10
* 20m pixel size:band5,band6,band7,band8a,band11,band12
* 10m pixel size:band2,band3,band4,band8

Upon first downloading the data, it is in the form of an array of arrays, where one sub-array represents a "row" of ground in the real world and for each sentinel 2 image, there is a separate file for each band (.jp2 files).<b2>
To use this data for machine learning purposes, we must transform it into a dataframe, where each row represents a pixel and each column a band. The values in the dataframe represent the measured reflectence in the given band.

```python
import rasterio
import numpy as np
import pandas as pd
import os
import glob
```

```python
# Find all JP2 files
folder = "C:\\Users\\mirianva\\Desktop\\sugis26\\Data science\\ids_PROJECT\\sentinel2_data\\S2B_poly8_9_12082022.SAFE\\GRANULE\\L2A_T35VME_A028376_20220812T093840\\IMG_DATA\\R60m"
files = glob.glob(os.path.join(folder, "*.jp2"))
```

```python
# select one file for an example. Here we choose the file for Band 1
band1 = rasterio.open(files[1], driver="JP2OpenJPEG")
band_array=band1.read()  # each value is pixel value describing the reflectance of a given band. Here are the band1 values for each pixel.
band_array #teh image is 1830 pixels wide and also 1830 pixels high
```

```python
# Now we convert all of the files into one dataframe:

# Dictionary to store the bands
bands = {}

for file in files:
    
    # Extract band name from filename
    band_name = os.path.basename(file).split("_")[-2].replace(".jp2", "")
    # print(band_name)
    
    # Read raster
    with rasterio.open(file) as src:
        data = src.read(1)

    # Flatten 2D array into 1D
    bands[band_name] = data.flatten()

# Create DataFrame
sentinel_df = pd.DataFrame(bands)
print(sentinel_df.head())
```

As we can see, the sentinel image does not contain any pixels that have "No data". That is very good. However, there are 380 pixels of dark areas (code 2), 908 pixels with possible cloud contamination (code 10) and 998 unclassified pixels (code7).

Let's look at the faulty pixels on an image:

```python
import rasterio
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap, BoundaryNorm
from matplotlib.patches import Patch

# Read JP2 file
with rasterio.open(files[12]) as src:
    image = src.read(1)

# Define pixel categories
categories = {
    2: "dark",
    4: "Forest",
    5: "bare",
    6: "water",
    7: "unclassified",
    10:"cloud"
}

# Define colors for each category
colors = [
    "red",
    "lightgrey",
    "lightgrey",
    "lightgrey",
    "green",
    "black"
]

# Create discrete colormap
cmap = ListedColormap(colors)

fig, ax = plt.subplots(figsize=(10, 10))

ax.imshow(image, cmap=cmap)
ax.set_title("Land cover classification")
ax.axis("off")

# Create custom legend
legend_elements = [
    Patch(facecolor=colors[i], label=label)
    for i, label in enumerate(categories.values())
]

ax.legend(
    handles=legend_elements,
    loc="upper right",
    title="Land cover classes"
)

plt.show()
```

We can clearly see on the image above, the pixels classified as cloud (value 10, color black) and dark area pixels (value 2, color red). It is very difficult to tell where the unclassified pixels are located.
